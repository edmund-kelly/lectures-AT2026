## A Function Call Always Returns a Value

* The execution of a `return` statement terminates the function call
* The function call also terminates when there are no more statements to execute
* If no expression follows `return` or there is no `return` statement, the function returns `None`       

````python
def get_larger(x, y):
    if x > y:
        return x
    if y > x:
        return y

ex1 = get_larger(3, 5)
ex2 = get_larger(6, 4)
ex3 = get_larger(3, 3)

print(ex1, ex2, ex3)
````

## Positional vs. Keyword Arguments

**Positional** arguments are matched to the parameters by their order. **Keyword** arguments (`name=value`) are matched by name, so their order does not matter.

````python
def describe_pet(name, animal):
    print(name, 'is a', animal)

describe_pet('Rex', 'dog')               # positional: matched by order
describe_pet(animal='dog', name='Rex')   # keyword: matched by name
describe_pet('dog', 'Rex')               # wrong order, wrong meaning
describe_pet('Rex', animal='dog')        # mixed: positional first
````

## Default Parameter Values

* Default values allow to call a function with fewer arguments than specified
* Default arguments cannot come before non-default arguments!

````python
def pretty_print(lst, sep, fullstop=True, capitalize=True):
    toprint = sep.join(lst)
    if fullstop:
        toprint += '.'
    if capitalize:
        toprint = toprint.capitalize()
    print(toprint)

wordlst = ['the', 'quick', 'brown', 'fox', 'jumps', 
           'over', 'the', 'lazy', 'dog']  # an English pangram

pretty_print(wordlst, ' ', True, True)
pretty_print(wordlst, ' ')
pretty_print(wordlst, ' ', False)
````

Using keyword arguments often improves legibility

````python
pretty_print(lst = wordlst, sep = ' ', fullstop = True, capitalize = True)
````

## Functions with Variable Number of Arguments

A parameter with a star, by convention `*args`, collects **any number of positional arguments** into a tuple. You already know such a function: `print()`.

````python
def double_many(*args):
    print('args:', args)   # all arguments, collected in a tuple
    return [2*i for i in args]

print(double_many(1))
print(double_many(1, 2, 3))
print(double_many())
````

## A Function Defines a New Scope

Each function call has its own **scope** (name space): its parameters and the variables it creates are **local**. They exist only while the function runs.

````python
def func(x, y):
    x = x + 1          # changes only the local x
    z = x + y          # z is a local variable
    print('inside:  x =', x)
    return z

x = 1
res = func(x, 5)
print('outside: x =', x)
# print(z)             # NameError: z only exists inside func
````

## The Global Scope

Variables defined outside any function are **global**. If a name is not defined inside a function, Python looks it up in the global scope, so functions can read global variables.

````python
GLOBVAR = 3            # global: defined outside any function

def print_global():
    # not defined in the function, so Python uses the global GLOBVAR
    print('inside:  GLOBVAR =', GLOBVAR)

print_global()
print('outside: GLOBVAR =', GLOBVAR)
````

## Other Ways to Import a Module

````python
import module as md               # import the module under a short name
md.my_func('Hello there!')
````

````python
from module import my_func        # import one function: no prefix needed
my_func('HELLO! DO YOU HEAR ME?')
````

Avoid `from module import *`: it imports every name, which can clash with names you already use.

## Functions Calling Themselves, A.K.A. Recursion

**Recursion** is a problem-solving method where the solution to a problem depends on solutions to smaller instances of the same problem.



````python
# Consider the sum of elements in a list 

lst = [1, 2, 3, 4, 5]
# We want (1 + 2 + 3 + 4 + 5), which is equivalent to (1 + (2 + (3 + (4 + 5)))).
# This suggests that we can reduce the problem to the problem of adding two numbers.

def list_sum(lst):
    """Assumes lst is a sequence of numeric types.
    Returns the sum of all elements in lst.
    """
    if len(lst) == 1:
        return lst[0]   # base case
    else:
        return lst[0] + list_sum(lst[1:])   # recursive case
    
print(list_sum(lst))
````

## Recursion Example: The Sum of a List

The sum of all elements in a list:

$$\operatorname{sum}([x_1, x_2, \ldots, x_n]) = x_1 + x_2 + \cdots + x_n, \qquad \text{e.g. } \operatorname{sum}([1, 2, 3, 4, 5]) = 15$$

Everything after the first element is itself the sum of a (shorter) list:

$$\operatorname{sum}([1, 2, 3, 4, 5]) = 1 + \underbrace{2 + 3 + 4 + 5}_{\operatorname{sum}([2, 3, 4, 5])} = 1 + \operatorname{sum}([2, 3, 4, 5])$$

Repeating this step, the list gets shorter until one element is left: $\operatorname{sum}([5]) = 5$

$$1 + \operatorname{sum}([2, 3, 4, 5]) = 1 + 2 + \operatorname{sum}([3, 4, 5]) = \cdots = 1 + 2 + 3 + 4 + \operatorname{sum}([5])$$

So, in general:

$$\operatorname{sum}([x_1, \ldots, x_n]) = \begin{cases} x_1 & \text{if } n = 1 & \text{(base case)} \\ x_1 + \operatorname{sum}([x_2, \ldots, x_n]) & \text{if } n > 1 & \text{(recursive case)} \end{cases}$$

## Recursion Example: The Factorial `n!`

The product of all positive integers less than or equal to $n$:

$$n! = n \cdot (n-1) \cdot (n-2) \cdots 2 \cdot 1, \qquad \text{e.g. } 5! = 5 \cdot 4 \cdot 3 \cdot 2 \cdot 1 = 120$$

Everything after the first factor is itself a factorial:

$$5! = 5 \cdot \underbrace{4 \cdot 3 \cdot 2 \cdot 1}_{4!} = 5 \cdot 4!$$

Repeating this step, the problem gets smaller until we reach $1! = 1$:

$$5! = 5 \cdot 4! = 5 \cdot 4 \cdot 3! = 5 \cdot 4 \cdot 3 \cdot 2! = 5 \cdot 4 \cdot 3 \cdot 2 \cdot 1!$$

So, in general:

$$n! = \begin{cases} 1 & \text{if } n = 1 & \text{(base case)} \\ n \cdot (n-1)! & \text{if } n > 1 & \text{(recursive case)} \end{cases}$$

## Recursion Example: What Is `n!`?

1. General case: `n! = n * (n-1)!`
2. Base case: `1! = 1`
3. Base case reached when `n==1`

````python
def factorial(n):
    """Assumes n is a positive integer.
    Returns n!.
    """
    if n == 1:
        return 1   # base case
    else:
        return n * factorial(n - 1)   # recursive case

print(factorial(5))
````

## Any Recursive Algorithm Can Be Transformed into an Iterative One!

Our recursive `list_sum`, rewritten as a loop:

<div class="code-columns"></div>

````python
def list_sum(lst):
    if len(lst) == 1:
        return lst[0]
    else:
        return lst[0] + list_sum(lst[1:])
````

````python
def list_sum_ite(lst):
    total = 0
    for x in lst:
        total = total + x
    return total
````

Each recursive call adds one element, and so does each pass through the loop. Instead of calls that wait for each other, the loop keeps the running sum in `total`.

## Comparing the Execution Time

````python
from time import time

nums = list(range(1, 901))   # [1, 2, ..., 900]

start_time = time()
list_sum(nums)
time_rec = time() - start_time
print('Recursion:', time_rec, 'seconds')

start_time = time()
list_sum_ite(nums)
time_ite = time() - start_time
print('Iteration:', time_ite, 'seconds')

print('The iterative version is', round(time_rec / time_ite),
      'times faster than the recursive one.')
````

The recursive `list_sum` does extra work: every call copies the rest of the list (`lst[1:]`), and Python has to keep track of all the calls that are still waiting.

## From Last Week: Approximation with Bisection Search

````python
def bisec_search(x, epsilon):
    """Assumes x and epsilon are numeric.
    Finds an approximation to the square root 
    of a number x using bisection search.
    """

    # Define interval for search
    low = 0
    high = max(1, x)

    # Start in the middle
    guess = (low + high) / 2 

    # Narrow down search interval until guess close enough
    while abs(guess**2 - x) >= epsilon:
        if guess**2 < x:
            low = guess
        else:
            high = guess
        guess = (low + high) / 2
        
    return guess

print(bisec_search(25, 0.01))
````

## Recursion Example: Approximation with Bisection Search

1. General case:
<pre><code>guess = (low + high) / 2
if guess**2 &lt; x: low = guess
else: high = guess</code></pre>
2. Base case: `return guess`
3. The base case is reached when `abs(guess**2 - x) < epsilon`

````python
def bisec_search_rec(low, high, x, epsilon):
    """Assumes low, high, x and epsilon are numeric.
    Finds an approximation within epsilon to the square root 
    of a number x using bisection search. 
    Uses a recursive implementation where low and high
    determine the search range.
    """
    
    # Start in the middle of the interval
    guess = (low + high) / 2
    
    # Stop when close enough
    if abs(guess**2 - x) < epsilon:
        return guess
    # Otherwise keep splitting
    else:
        if guess**2 < x:
            return bisec_search_rec(guess, high, x, epsilon)
        else:
            return bisec_search_rec(low, guess, x, epsilon)

print(bisec_search_rec(0, 25, 25, 0.01))
````

## Recursion Example: What Is the n-th Fibonacci Number?

1. General case: `fib(n) = fib(n - 1) + fib(n - 2)` 

2. Base case: `fib(0) = 0`, `fib(1) = 1`

3. The base case is reached when `n == 0` or `n == 1`

````python
def fib(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fib(n - 1) + fib(n - 2)

for i in range(10): 
    print(fib(i))
````

## The Fibonacci Numbers: Recursion vs. Iteration

Even if you can formulate a problem in recursive terms, it does not mean that recursion is the most efficient way to solve it.

````python
def fib_rec(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fib_rec(n - 1) + fib_rec(n - 2)


def fib_ite(n):    
    first = 0
    second = 1
    for i in range(n):
        old_first = first  # remember what first was
        first = second  # shift the first forward
        second = old_first + first  # estimate the new second
    return first
````

## The Fibonacci Numbers: Comparing the Execution Time

````python
from time import time

start_time = time()
fib33_rec = fib_rec(33)
print('Recursion:', (time() - start_time), 'seconds')

start_time = time()
fib33_ite = fib_ite(33)
print('Iteration:', (time() - start_time), 'seconds')

print(fib33_rec == fib33_ite) 
````

An alternative way to do this is with the `timeit` module. By default, it runs the code snippet 1,000,000 times and then returns the execution time.

````python
from timeit import timeit

print('Recursion:', timeit("fib_rec(6)", setup="from __main__ import fib_rec"))
print('Iteration:', timeit("fib_ite(6)", setup="from __main__ import fib_ite"))
````
