# How Functions Work

## Defining and Calling Functions

**Defining** a function:

<pre><code>def <i>function_name</i>(<i>list of parameters</i>):
    <i>body of function</i></code></pre>

**Calling** a function:

<pre><code><i>function_name</i>(<i>arguments</i>)</code></pre>

* Functions can take 0 or more arguments and return 1 or more values
* If a function does not have a `return` statement, it returns `None`

## Functions Take Arguments by Reference (1/2)

A function receives the object you pass, not a copy. If the function **changes** a mutable object (e.g. `append` to a list), you see the change outside the function:

```python
def change_list(alist):
    alist.append(0)

mylist = [1, 2, 3]
change_list(mylist)
print(mylist)
```

## Functions Take Arguments by Reference (2/2)

But if the function **assigns** a new object to the parameter, only the local name points to it; the list outside stays the same:

```python
def zero_list(alist):
    """Takes a list and returns another list of the same length 
    that looks like [0, 0, 0, ...].
    """
    alist = [0]*len(alist)  # Creates a new local reference for alist

mylist = [1, 2, 3]
zero_list(mylist)
print(mylist)
```

## Using Functions to Improve Legibility and Modularity

Functions with descriptive names make code easier to read. Consider this code:

```python
# You are given two points in 2-D space
x = (1, 1)
y = (5, 4)

# Calculate the area of the circle if one of the points is
# the circle center and the other is on the perimeter and
# then calculate the side of the square with the same area
r_sq = (x[0] - y[0])**2 + (x[1] - y[1])**2
area = 3.14*r_sq
sq_side = area**0.5
print(sq_side)
```

## Using Functions Inside List Comprehensions

You can call your own functions inside a list comprehension:

```python
def sq_or_sqrt(x):
    """Assumes x is numeric. Returns the square of x if x is negative
    and the square root of x if x is nonnegative.
    """
    if x < 0:
        return x**2
    else:
        return x**0.5

lst = [sq_or_sqrt(i) for i in range(-5, 6)]
print(lst)
```
